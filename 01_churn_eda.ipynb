```python
print('\nChurn Distribution Visualization :')
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.show()
```

```python
print("\nTenure vs Churn :")
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Churn", y="tenure")

plt.title("Tenure vs Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.show()
```

```python
print("\nMonthly Charges vs Churn :")
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Churn", y="MonthlyCharges")

plt.title("Monthly Charges vs Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")

plt.show()
```

```python
print('\nContract vs Churn :')
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Contract Type vs Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")

plt.show()
```

```python
print("\nCategorical Churn Analysis :")
categorical_features = [
    "InternetService",
    "PaymentMethod",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "PaperlessBilling"
]

for feature in categorical_features:
    print(f"\n{'='*60}")
    print(f"{feature} vs Churn")
    print(f"{'='*60}")

    result = pd.crosstab(
        df[feature],
        df["Churn"],
        normalize="index"
    ) * 100

    print(result.round(2))
```

```python
print("\nNumeric Churn Analysis :")
numeric_features = ["tenure", "MonthlyCharges", "TotalCharges"]

print("Numeric Features vs Churn")
print("=" * 60)

for feature in numeric_features:
    print(f"\n{feature} vs Churn")
    print("-" * 40)

    result = df.groupby("Churn")[feature].agg(
        ["count", "mean", "median", "min", "max"]
    )

    print(result.round(2))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv("data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
print("Shape:", df.shape)
print(df.info())
```

```python
churn_rate = df["Churn"].value_counts(normalize=True) * 100
print(churn_rate.round(2))
```

```python
pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
).mul(100).round(2)
```

```python
pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
).mul(100).round(2)
```

```python
pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
).mul(100).round(2)
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"]
)

pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
).mul(100).round(2)
```

```python
df["MonthlyChargeGroup"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0, 40, 70, 100, 120],
    labels=["Low", "Medium", "High", "Very High"]
)

pd.crosstab(
    df["MonthlyChargeGroup"],
    df["Churn"],
    normalize="index"
).mul(100).round(2)
```

```python
print("Total Customers:", len(df))
print("Churned Customers:", (df["Churn"] == "Yes").sum())
print("Churn Rate:", round((df["Churn"] == "Yes").mean() * 100, 2), "%")
print("Average Monthly Charges:", round(df["MonthlyCharges"].mean(), 2))
print("Average Tenure:", round(df["tenure"].mean(), 2))
```

```python
df.to_csv("data/churn_business_analysis.csv", index=False)
```
